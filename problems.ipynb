## Problem 1: Get Some Data


Problem 1 involves using *yfinance* to download historical data from NVIDIA, Broadcom, Arm, AMD and IBM. 
Writing a short function to create a folder called data and storing the downloaded data into a CSV file. The filename uses the format YYYYMMDD-HHmmss.csv, so it records the date and time when the data was downloaded.

```python
# Function to download and save historical stock data.
def get_data():

    #Download historical stock data over the last 5 days with 1-hour intervals.
    data = yf.download(["NVDA", "AVGO", "ARM", "AMD", "IBM"], period="5d", interval="1h")

    # Set up the data folder.
    data_folder = Path("data")
    data_folder.mkdir(exist_ok=True)

    # Create a filename using the current date and time.
    timestamp = datetime.now().strftime("%Y%m%d-%H%M%S")
    filename = data_folder / f"{timestamp}.csv"

    # Save the downloaded data to a CSV file.
    data.to_csv(filename, index=False)

    #Print the filename where the data is saved.
    print(f"Data saved to {filename}")
```

```python
# Call the function to download and save the data.
get_data()
```
